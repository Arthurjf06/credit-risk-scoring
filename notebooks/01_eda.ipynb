```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/cs-training.csv", index_col=0)
print(df.shape)
df.head()
```

```python
df.info()
df.describe().T
```

```python
alvo = "SeriousDlqin2yrs"
print(df[alvo].value_counts())
print(df[alvo].value_counts(normalize=True).round(4))
```

```python
ausentes = df.isna().mean().sort_values(ascending=False)
ausentes[ausentes > 0].round(4)
```

```python
df.hist(bins=40, figsize=(14, 10))
plt.tight_layout()
plt.show()
```

```python
df["faixa_idade"] = pd.cut(df["age"], bins=[0, 25, 35, 45, 55, 65, 120])
print(df.groupby("faixa_idade", observed=True)[alvo].mean().round(4))

df["faixa_renda"] = pd.qcut(df["MonthlyIncome"], q=5, duplicates="drop")
print(df.groupby("faixa_renda", observed=True)[alvo].mean().round(4))
```

```python
numericas = df.select_dtypes("number")
numericas.corr()[alvo].drop(alvo).sort_values()
```
